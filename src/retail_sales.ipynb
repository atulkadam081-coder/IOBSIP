```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_store_sales.csv")
df.head()
```

```python
df.columns
```

```python
df.groupby('Category')['Unit_Price'].sum().sort_values(ascending= False).plot(kind= 'bar', color= 'green')
plt.title('Total Amount By Category')
plt.xlabel('Category')
plt.ylabel('Unit_Price')
plt.show()
```

```python
sns.scatterplot(data= df, x= 'Cost', y= 'Profit', color= 'orange')
plt.title('Cost and Profit')
plt.show()
```

```python
df.columns
```

```python
df.groupby('Payment_Method')['Unit_Price'].sum().plot(kind= 'pie' , autopct= '%1.1f%%' )
plt.title("Amount By Payment Method")
plt.show()
```
